```python
import numpy as np
from data_tools.db import Manager
from sympy import Symbol
from sympy.solvers import solve

from mff.new_cap import generate_base_data, read_base_data

engine = Manager("mvh-admin", "mvh").engine
```

```python
generate_base_data(2024, engine)
```

```python
data = read_base_data(2024)
```

```python
def find_criss_area_t1(biss_ter):
    if biss_ter <= 1200 and biss_ter != 0:
        return min(10, biss_ter)
    else:
        return 0


def find_criss_area_t2(biss_ter):
    if biss_ter <= 1200 and biss_ter != 0:
        return min(140, max(biss_ter - 10, 0))
    else:
        return 0


area1 = data["area_biss_criss"].apply(find_criss_area_t1).sum()
area2 = data["area_biss_criss"].apply(find_criss_area_t2).sum()

x = Symbol("x")

res = float(solve(2 * x * area1 + x * area2 - 189176282.78, x)[0])
redist_per_ha = [2 * res, res]
redist_per_ha
```
