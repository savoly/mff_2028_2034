```python
import numpy as np
import pandas as pd

from mff.new_cap import read_extended_base_data
```

```python
data = read_extended_base_data(2026)
```

```python
def calc_sfs_basic(data, limit, area_per_subs):
    data = data[
        (data["area_biss_criss"] > 0) & (data["area_biss_criss"] <= limit)
    ].copy()

    terulet = data["area_biss_criss"].sum()
    terulet_jogosult = np.ceil(data["area_biss_criss"]).sum()

    return pd.DataFrame(
        {
            "Felső határ, ha": [f"{limit} ha"],
            "Üzemszám, db": [data["regszam"].nunique()],
            "Terület, ha": [terulet],
            "Jogosult terület, ha": [terulet_jogosult],
            "Igény, millió EUR": [terulet_jogosult * area_per_subs / 1e6],
        }
    )
```

```python
def calc_sfs(data, limit, area_per_subs, area_per_subs_per_dabis):
    data = data[(data["area_biss_criss"] > 0) & (data["area_biss_criss"] <= limit)]

    data = data[
        (
            data["area_biss_criss"] * area_per_subs_per_dabis
            + data["area_tk_cukorrepa"] * 715.03
        )
        < (np.ceil(data["area_biss_criss"]) * area_per_subs)
    ]

    terulet = data["area_biss_criss"].sum()

    return pd.DataFrame(
        {
            "uzszam": [data["regszam"].nunique()],
            "terulet": [terulet],
            "igény, millió EUR": [(np.ceil(terulet) * area_per_subs) / 1e6],
        }
    )
```
