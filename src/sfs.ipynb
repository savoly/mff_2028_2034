```python
import numpy as np
import pandas as pd

from mff.new_cap import read_extended_base_data

data = read_extended_base_data(2026)


def calc_sfs_upper(data, limit, area_per_subs):
    data = data[
        (data["area_biss_criss"] > 0) & (data["area_biss_criss"] <= limit)
    ].copy()

    terulet = data["area_biss_criss"].sum()
    terulet_jogosult = np.ceil(data["area_biss_criss"]).sum()

    return pd.DataFrame(
        {
            "Felső határ, ha": [f"{limit} ha"],
            "Üzemszám, db": [data["regszam"].nunique()],
            "Terület, ha": [terulet],
            "Jogosult terület, ha": [terulet_jogosult],
            "Igény, millió EUR": [terulet_jogosult * area_per_subs / 1e6],
        }
    )


def calc_sfs_lower(data, limit, area_per_subs):
    data = data[(data["area_biss_criss"] > 0) & (data["area_biss_criss"] <= limit)]

    cis_cols = [
        "area_tk_cukorrepa",
        "area_tk_szemes_feherjenoveny",
        "area_tk_szalas_feherjenoveny",
        "area_tk_extenziv_gyumolcs",
        "area_tk_intenziv_gyumolcs",
        "area_tk_ipari_olajnoveny",
        "area_tk_ipari_zoldsegnoveny",
        "area_tk_zoldsegnoveny",
        "area_tk_rizs",
        "count_tk_hizottbika",
        "count_tk_anyatehen",
        "count_tk_tejhasznu_tehen",
        "count_tk_anyajuh",
    ]

    mask = ~(data[cis_cols] > 0).any(axis=1)
    data = data[mask]

    terulet = data["area_biss_criss"].sum()
    terulet_jogosult = np.ceil(data["area_biss_criss"]).sum()

    return pd.DataFrame(
        {
            "Felső határ, ha": [f"{limit} ha"],
            "Üzemszám, db": [data["regszam"].nunique()],
            "Terület, ha": [terulet],
            "Jogosult terület, ha": [terulet_jogosult],
            "Igény, millió EUR": [terulet_jogosult * area_per_subs / 1e6],
        }
    )
```

```python
sfs_upper = pd.concat(
    [
        calc_sfs_upper(data, 5, 200),
        calc_sfs_upper(data, 10, 200),
        calc_sfs_upper(data, 15, 200),
        calc_sfs_upper(data, 20, 200),
        calc_sfs_upper(data, 23, 200),
    ]
)

sfs_lower = pd.concat(
    [
        calc_sfs_lower(data, 5, 200),
        calc_sfs_lower(data, 10, 200),
        calc_sfs_upper(data, 15, 200),
        calc_sfs_lower(data, 20, 200),
        calc_sfs_lower(data, 23, 200),
    ]
)
```
